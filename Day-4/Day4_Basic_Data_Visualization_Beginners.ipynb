## 1. Import Libraries

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

## 2. Create a Small Dataset

We will use a simple sales dataset so beginners can understand every column.

````python
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 14000, 18000, 21000, 19000],
    "Profit": [2000, 2500, 2200, 3200, 4000, 3500],
    "Category": ["Phone", "Laptop", "Phone", "Laptop", "Tablet", "Laptop"],
    "Region": ["North", "South", "North", "East", "West", "South"]
}

df = pd.DataFrame(data)
df
````

## 3. Understand the Data

````python
print(df.shape)
print(df.info())
````

````python
df.describe(numeric_only=True)
````

## 4. Line Chart

**Use a line chart for trends over time.**

````python
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
````

## 5. Bar Chart

**Use a bar chart to compare categories.**

````python
plt.bar(df["Month"], df["Sales"])
plt.title("Sales by Month")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
````

## 6. Histogram

**Use a histogram to understand the distribution of values.**

````python
plt.hist(df["Sales"])
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
````

## 7. Scatter Plot

**Use a scatter plot to see the relationship between two numerical columns.**

````python
plt.scatter(df["Sales"], df["Profit"])
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
````

# Seaborn

Seaborn is built on Matplotlib and makes statistical charts easier to create.

````python
sns.set_style("whitegrid")
````

## 8. Seaborn Barplot

````python
sns.barplot(data=df, x="Category", y="Sales")
plt.title("Sales by Category")
plt.show()
````

## 9. Countplot

**Countplot shows how many records belong to each category.**

````python
sns.countplot(data=df, x="Region")
plt.title("Transactions by Region")
plt.show()
````

## 10. Boxplot

A boxplot helps identify the spread of data and possible outliers.

````python
sns.boxplot(data=df, y="Sales")
plt.title("Sales Boxplot")
plt.show()
````

## 11. Heatmap

A heatmap can show correlation between numerical columns.

````python
corr = df[["Sales", "Profit"]].corr()

sns.heatmap(corr, annot=True)
plt.title("Sales and Profit Correlation")
plt.show()
````

# 12. Four Charts Together

Create a simple 2 × 2 visual summary:

1. Sales trend
2. Sales distribution
3. Sales comparison
4. Sales vs Profit

````python
fig, ax = plt.subplots(2, 2, figsize=(10, 7))

ax[0, 0].plot(df["Month"], df["Sales"], marker="o")
ax[0, 0].set_title("Sales Trend")

ax[0, 1].hist(df["Sales"])
ax[0, 1].set_title("Sales Distribution")

ax[1, 0].bar(df["Month"], df["Sales"])
ax[1, 0].set_title("Sales Comparison")

ax[1, 1].scatter(df["Sales"], df["Profit"])
ax[1, 1].set_title("Sales vs Profit")

plt.tight_layout()
plt.show()
````

# 15. Chart Selection – Quick Revision

| Question | Chart |
|---|---|
| Show a trend | Line chart |
| Compare values | Bar chart |
| Show distribution | Histogram |
| Show relationship | Scatter plot |
| Find possible outliers | Boxplot |
| Show correlation | Heatmap |
| Count categories | Countplot |

### Remember

```text
Trend        → Line
Comparison   → Bar
Distribution → Histogram
Relationship → Scatter
Outliers     → Boxplot
Correlation  → Heatmap
```
