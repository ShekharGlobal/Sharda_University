```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
#Mean, Median, Mode
from statistics import mean, median, mode

data = [10, 20, 20, 30, 40]

print("Mean:", mean(data))
print("Median:", median(data))
print("Mode:", mode(data))
```

```python
#Variance & Standard Deviation
import numpy as np

data = [10, 20, 30, 40, 50]

print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))
```

```python

#Quartiles & IQR
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Q1:", q1)
print("Median:", q2)
print("Q3:", q3)
print("IQR:", q3 - q1)
```

```python
#Five-Number Summary
import pandas as pd

data = pd.Series([10, 20, 30, 40, 50, 60, 70])

print(data.describe())
```

```python
#Box Plot & Outlier
data = [10, 20, 30, 40, 50]

plt.boxplot(data)

plt.title("My First Box Plot")
plt.show()
```

```python
#Z-Score
#Z-score tells us how far a value is from the mean in terms of standard deviations.
from scipy.stats import zscore

data = [10, 20, 30, 40, 50]

print(zscore(data))
```

```python
#Skewness
#Positive → tail to right
#Negative → tail to left
# 0        → roughly symmetric

import pandas as pd

data = pd.Series([10, 20, 20, 30, 40, 100])

print("Skewness:", data.skew())
```

```python
#Correlation
#Here we are checking the relationship between Hours and Marks.
import pandas as pd

data = {
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

print(df.corr())
```

```python
#Probability
#Simple Probability
# 30
favourable = 3
total = 10

probability = favourable / total

print("Probability:", probability)
```

```python

import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

plt.hist(data)
plt.title("Normal Distribution")
plt.show()
```

```python
#Binomial Distribution
#Example: 10 coin tosses
import numpy as np

data = np.random.binomial(10, 0.5, 10)

print(data)
```

```python
#Poisson Distribution
#Example: customers arriving at a shop.
import numpy as np

data = np.random.poisson(5, 10)

print(data)
```

```python
#Two-Sample t-test
from scipy.stats import ttest_ind

group1 = [60, 65, 70, 72, 68]
group2 = [75, 80, 78, 82, 79]

result = ttest_ind(group1, group2)

print(result)
```

```python
#Chi-Square
from scipy.stats import chi2_contingency

data = [
    [30, 20],
    [15, 35]
]

result = chi2_contingency(data)

print("p-value:", result[1])
```

```python
#ANOVA
from scipy.stats import f_oneway

a = [70, 72, 75]
b = [80, 82, 85]
c = [90, 88, 92]

result = f_oneway(a, b, c)

print(result)
```

```python
#EDA — Exploratory Data Analysis
#Create Dataset
import pandas as pd

data = {
    "Name": ["Amit", "Ravi", "Neha", "Priya", "John"],
    "Marks": [80, 70, 90, 60, 75],
    "Hours": [5, 3, 6, 2, 4]
}

df = pd.DataFrame(data)

print(df)
```

```python
#Understand the Dataset
print(df.head())
print(df.shape)
print(df.columns)
print(df.info())
```

```python
#Statistical Summary

print(df.describe())
```

```python
#Check Missing Values
print(df.isnull().sum())
```

```python
#Univariate Analysis
#Analyze one variable
#What does the Marks column look like?
print(df["Marks"].mean())

df["Marks"].hist()

plt.show()
```

```python
#Bivariate Analysis
#Analyze two variables.
#Does studying more hours relate to higher marks?
plt.scatter(df["Hours"], df["Marks"])

plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
#Multivariate Analysis
#Analyze multiple variables together.
print(df.corr(numeric_only=True))
```

```python
#EDA Outlier Analysis
plt.boxplot(df["Marks"])

plt.title("Marks Outliers")

plt.show()
```
